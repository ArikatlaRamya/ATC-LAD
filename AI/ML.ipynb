```python
#lab 1
# Google Colab - PIN Generator

# List of allowed even digits
even_digits = [0, 2, 4, 6, 8]

count = 0

print("Valid 4-digit PINs:\n")

# Generate all possible PINs
for d1 in even_digits:
    for d2 in even_digits:
        for d3 in even_digits:
            for d4 in even_digits:
                # Check if the sum of digits is 16
                if d1 + d2 + d3 + d4 == 16:
                    pin = f"{d1}{d2}{d3}{d4}"
                    print(pin)
                    count += 1

print("\n----------------------------")
print("Total valid PINs:", count)
```

```python
#lab 2
# AI & ML Lab Practical
# BFS and DFS on a Social Network Graph
# Google Colab Code

from collections import deque

# Graph Representation
graph = {
    "Alice": ["Charlie", "David"],
    "Charlie": ["Alice", "Emma"],
    "David": ["Alice", "Emma", "Fred"],
    "Emma": ["Bob", "Charlie", "David"],
    "Fred": ["Bob", "David"],
    "Bob": ["Emma", "Fred"]
}

# Arrange neighbors alphabetically
for node in graph:
    graph[node] = sorted(graph[node])


# ---------------- BFS ----------------
def bfs(graph, start, goal):
    queue = deque([(start, [start])])
    visited = set()

    print("===== Breadth First Search (BFS) =====\n")

    while queue:
        current, path = queue.popleft()

        if current not in visited:
            visited.add(current)

            print("Current Node :", current)
            print("Queue        :", [item[0] for item in queue])
            print("Visited      :", list(visited))
            print()

            if current == goal:
                print("Goal Found!")
                return path

            for neighbor in graph[current]:
                if neighbor not in visited:
                    queue.append((neighbor, path + [neighbor]))

    return None


# ---------------- DFS ----------------
def dfs(graph, start, goal):
    stack = [(start, [start])]
    visited = set()

    print("\n===== Depth First Search (DFS) =====\n")

    while stack:
        current, path = stack.pop()

        if current not in visited:
            visited.add(current)

            print("Current Node :", current)
            print("Stack        :", [item[0] for item in stack])
            print("Visited      :", list(visited))
            print()

            if current == goal:
                print("Goal Found!")
                return path

            # Reverse order so alphabetical order is maintained
            for neighbor in reversed(graph[current]):
                if neighbor not in visited:
                    stack.append((neighbor, path + [neighbor]))

    return None


# Source and Goal
source = "Alice"
goal = "Bob"

# Run BFS
bfs_path = bfs(graph, source, goal)
print("BFS Path :", " -> ".join(bfs_path))

# Run DFS
dfs_path = dfs(graph, source, goal)
print("DFS Path :", " -> ".join(dfs_path))
```
